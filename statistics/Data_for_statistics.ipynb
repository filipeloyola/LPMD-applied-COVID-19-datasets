```python
import pandas as pd
```

```python
df = pd.read_excel("todos_resultados_instantaneos.xlsx")

df
```

```python
df["Métrica"].unique()
```

```python
grupos = df.groupby("Métrica")

df_AUC = grupos.get_group("AUC")
df_recall = grupos.get_group("Recall")
df_spe = grupos.get_group("Specificity")

display(df_AUC)
display(df_recall)
display(df_spe)
```

```python
df_AUC_1 = df_AUC[["Repetition", "Dataset", "Imputer", "Métrica", "SVM"]]
df_AUC_1 = df_AUC_1.rename(columns={"SVM": "score"})
display(df_AUC_1)

df_AUC_2 = df_AUC[["Repetition", "Dataset", "Imputer", "Métrica", "RF"]]
df_AUC_2 = df_AUC_2.rename(columns={"RF": "score"})
display(df_AUC_2)

df_AUC_3 = df_AUC[["Repetition", "Dataset", "Imputer", "Métrica", "GB"]]
df_AUC_3 = df_AUC_3.rename(columns={"GB": "score"})
display(df_AUC_3)
```

```python
df_AUC = pd.concat([df_AUC_1, df_AUC_2, df_AUC_3],  ignore_index=True)
df_AUC.head()
```

```python
df_AUC_pivot = df_AUC.pivot_table(index=["Repetition", "Dataset", "Métrica"], columns="Imputer", values="score")

df_AUC_pivot
```

```python
df_AUC_pivot_reset = df_AUC_pivot.reset_index()
df_AUC_pivot_reset.columns = ['_'.join(col).strip() if isinstance(col, tuple) else col for col in df_AUC_pivot_reset.columns.values]
df_AUC_pivot_reset
```

```python
df_AUC = df_AUC_pivot_reset[['KNN', 'LPMD', 'LPMD2', 'MICE', 'Mean', 'PMIVAE', 'SAEI']]
df_AUC
```

```python
df_AUC.to_csv("df_AUC_real_datasets.csv", index=False)
```

# Refazendo procedimento para RECALL

```python
df_rec_1 = df_recall[["Repetition", "Dataset", "Imputer", "Métrica", "SVM"]]
df_rec_1 = df_rec_1.rename(columns={"SVM": "score"})
display(df_rec_1)

df_rec_2 = df_recall[["Repetition", "Dataset", "Imputer", "Métrica", "RF"]]
df_rec_2 = df_rec_2.rename(columns={"RF": "score"})
display(df_rec_2)

df_rec_3 = df_recall[["Repetition", "Dataset", "Imputer", "Métrica", "GB"]]
df_rec_3 = df_rec_3.rename(columns={"GB": "score"})
display(df_rec_3)

df_rec = pd.concat([df_rec_1, df_rec_2, df_rec_3],  ignore_index=True)
display(df_rec)

df_rec_pivot = df_rec.pivot_table(index=["Repetition", "Dataset", "Métrica"], columns="Imputer", values="score")
df_rec_pivot_reset = df_rec_pivot.reset_index()
df_rec_pivot_reset.columns = ['_'.join(col).strip() if isinstance(col, tuple) else col for col in df_rec_pivot_reset.columns.values]
df_rec = df_AUC_pivot_reset[['KNN', 'LPMD', 'LPMD2', 'MICE', 'Mean', 'PMIVAE', 'SAEI']]
df_rec
```

```python
df_rec.to_csv("df_REC_real_datasets.csv", index=False)
```

# Refazendo procedimento para Specificity

```python
df_spe_1 = df_spe[["Repetition", "Dataset", "Imputer", "Métrica", "SVM"]]
df_spe_1 = df_spe_1.rename(columns={"SVM": "score"})
display(df_spe_1)

df_spe_2 = df_spe[["Repetition", "Dataset", "Imputer", "Métrica", "RF"]]
df_spe_2 = df_spe_2.rename(columns={"RF": "score"})
display(df_spe_2)

df_spe_3 = df_spe[["Repetition", "Dataset", "Imputer", "Métrica", "GB"]]
df_spe_3 = df_spe_3.rename(columns={"GB": "score"})
display(df_spe_3)

df_spe = pd.concat([df_spe_1, df_spe_2, df_spe_3],  ignore_index=True)
display(df_spe)

df_spe_pivot = df_spe.pivot_table(index=["Repetition", "Dataset", "Métrica"], columns="Imputer", values="score")
df_spe_pivot_reset = df_spe_pivot.reset_index()
df_spe_pivot_reset.columns = ['_'.join(col).strip() if isinstance(col, tuple) else col for col in df_spe_pivot_reset.columns.values]
df_spe = df_spe_pivot_reset[['KNN', 'LPMD', 'LPMD2', 'MICE', 'Mean', 'PMIVAE', 'SAEI']]
df_spe
```

```python
df_spe.to_csv("df_SPE_real_datasets.csv", index=False)
```

# Juntando as três métricas para aumentar o volume de dados

```python
df_real_datasets = pd.concat([df_AUC, df_rec, df_spe], ignore_index=True)

df_real_datasets
```
